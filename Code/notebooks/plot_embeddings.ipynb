```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

data = np.load("../embeddings/emb_agnews_sbert.npz")

X = data["embeddings"]
y = data["labels"]

print("Embeddings:", X.shape)
print("Labels:", y.shape)
print("Number of classes:", len(np.unique(y)))
print("Classes:", np.unique(y))

# Reduce 768D embeddings to 2D
X_2d = PCA(n_components=2).fit_transform(X)

# Get unique classes
classes = np.unique(y)

# 20 distinct colors
cmap = plt.get_cmap("tab20", len(classes))

plt.figure(figsize=(14, 9))

for i, cls in enumerate(classes):
    mask = y == cls

    plt.scatter(
        X_2d[mask, 0],
        X_2d[mask, 1],
        s=6,
        alpha=0.6,
        color=cmap(i),
        label=f"Class {cls}"
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("SBERT Embeddings — 20 Newsgroups")

plt.legend(
    title="Classes",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    fontsize=8
)

plt.tight_layout()
plt.show()
```
